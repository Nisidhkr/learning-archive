## Pandas Series

a series is a one dimensional labeled array capable of holding any data type. the axis lables are collectively called the index.

## Creating Series 


There are multiple way to creaate pandas Series:

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
labeles = ['a','b','c']
list = [10,20,30]
arr = np.array([10,20,30])
d = {1:10,2:20,3:30}
```

```python
pd.Series(list, index=labeles)
```

```python
pd.Series(arr, index=labeles)
```

```python
pd.Series(d)
```
