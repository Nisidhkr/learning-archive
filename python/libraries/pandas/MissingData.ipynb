### Finding Missing data

```python
import numpy as np 
import pandas as pd 
```

```python
data = {
'A': [1, 2, np.nan, 4, 5],
'B': [np.nan, 2, 3, 4, 5],
'C': [1, 2, 3, np.nan, np.nan],
'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
df.isna().sum()

```

```python
df.isna().any()
```

## Removing Missing Data

```python
df.dropna()
```

```python
df.dropna(axis=1)
```

```python
df.dropna(thresh=3)
```

```python
df.dropna(thresh=4)
```

```python
df.fillna(0)
```

```python
value = {'A':50, 'B':0, 'C':200, 'D':10}
```

```python
df.fillna(value=value)
```

```python
df.fillna(df.mean())
```
