# Data Frames

when we combine multiple series together  we get a data frame 

```python
import numpy as np
import pandas as pd
```

Creating Data Frames


```python
data = {
'Name': ['John', 'Anna', 'Peter', 'Linda'],
'Age': [28, 34, 29, 42],
'City': ['New York', 'Paris', 'Berlin', 'London'],
'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
```

```python
df_list = [
['John', 28, 'New York', 65000],
['Anna', 34, 'Paris', 70000],
['Peter', 29, 'Berlin', 62000],
['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(df_list)
```

## Conditional Statement 

```python
df[df['Age'] > 30]  # Filter rows based on condition
```

```python
df[(df['Age'] > 30) & (df["City"] == "London")] # Filter rows where Name is Linda and Age is greater than 30
```
